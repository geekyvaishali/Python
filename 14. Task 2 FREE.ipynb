### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python

# Take annual CTC in lakhs
ctc = float(input("Enter your CTC in lakhs: "))

# Calculate HRA, DA and PF
hra = ctc * 10 / 100
da = ctc * 5 / 100
pf = ctc * 3 / 100

# Calculate tax according to CTC
if ctc < 5:
    tax = 0
elif ctc <= 10:
    tax = ctc * 10 / 100
elif ctc <= 20:
    tax = ctc * 20 / 100
else:
    tax = ctc * 30 / 100

# Calculate annual and monthly in-hand salary
annual_salary = ctc - hra - da - pf - tax
monthly_salary = annual_salary * 100000 / 12

# Display result
print("HRA (lakhs):", hra)
print("DA (lakhs):", da)
print("PF (lakhs):", pf)
print("Tax (lakhs):", tax)
print("Monthly in-hand salary: ₹", round(monthly_salary, 2))
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python

# Take input from the user
n = int(input("Enter a number: "))

# Initialize factorial
fact = 1

# Calculate factorial using loop
for i in range(1, n + 1):
    fact = fact * i

# Print the result
print(fact)
````

### `Problem 7` - Reverse a given integer number.

Example:

`Input:`

```bash
76542
```

`Output:`

```bash
24567
```

````python

# Take input from the user
n = int(input("Enter a number: "))

# Initialize reverse
rev = 0

# Reverse the number
while n > 0:
    rem = n % 10
    rev = rev * 10 + rem
    n = n // 10

# Print the result
print(rev)
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash
30
```

`Output:`

```bash
276
```

````python

n = int(input("Enter a number: "))

i = 1
total = 0

while i <= n:
    if i % 5 == 0:
        i += 1
        continue

    total = total + i

    if total > 300:
        total = total - i
        break

    i += 1

print(total)
````

### `Problem 9`: Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python

total = 0
count = 0

while True:
    n = float(input("Enter a number (0 to stop): "))

    if n == 0:
        break

    total = total + n
    count = count + 1

print("Sum:", total)

if count > 0:
    average = total / count
    print("Average:", average)
else:
    print("No numbers entered.")
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python

import math

x = 0
y = 0

moves = """UP 5
DOWN 3
LEFT 3
RIGHT 2"""

for move in moves.splitlines():
    direction, steps = move.split()
    steps = int(steps)

    if direction == "UP":
        y += steps
    elif direction == "DOWN":
        y -= steps
    elif direction == "LEFT":
        x -= steps
    elif direction == "RIGHT":
        x += steps

distance = math.sqrt(x**2 + y**2)
print(round(distance))
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python

start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

for num in range(start, end + 1):
    temp = num
    total = 0

    while temp > 0:
        digit = temp % 10
        total = total + digit ** 3
        temp = temp // 10

    if total == num:
        print(num, end=" ")
        
if start <= 0 <= end:
    print(0, end=" ")
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python

import math

H = int(input("Enter hour: "))
M = int(input("Enter minutes: "))

hour_angle = (H % 12) * 30 + M * 0.5
minute_angle = M * 6

angle = abs(hour_angle - minute_angle)

if angle > 180:
    angle = 360 - angle

print(math.floor(angle))
````

###`Problem 15`:Given two rectangles, find if the given two rectangles overlap or not. A rectangle is denoted by providing the x and y coordinates of two points: the left top corner and the right bottom corner of the rectangle. Two rectangles sharing a side are considered overlapping. (L1 and R1 are the extreme points of the first rectangle and L2 and R2 are the extreme points of the second rectangle).

Note: It may be assumed that the rectangles are parallel to the coordinate axis.

<img src='https://www.geeksforgeeks.org/wp-content/uploads/rectanglesOverlap.png' width='300' height='200'>

````python

# Input coordinates of rectangle 1
x1, y1 = map(int, input("Enter left top coordinates of rectangle 1: ").split())
x2, y2 = map(int, input("Enter right bottom coordinates of rectangle 1: ").split())

# Input coordinates of rectangle 2
x3, y3 = map(int, input("Enter left top coordinates of rectangle 2: ").split())
x4, y4 = map(int, input("Enter right bottom coordinates of rectangle 2: ").split())

# Check whether rectangles overlap
if x1 > x4 or x3 > x2 or y1 < y4 or y3 < y2:
    print("Rectangles do not overlap")
else:
    print("Rectangles overlap")
````
