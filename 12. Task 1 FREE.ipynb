### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python

# Take two numbers as input
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Swap the numbers using a temporary variable
temp = a
a = b
b = temp

# Print the swapped numbers
print("After swapping:")
print("First number:", a)
print("Second number:", b)
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python

import math

# Take first coordinate as input
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

# Take second coordinate as input
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

# Calculate Euclidean distance
distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Print the result
print("Euclidean distance:", distance)
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python

# Take input from the user
P = float(input("Enter principal amount: "))
R = float(input("Enter rate of interest: "))
T = float(input("Enter time period in years: "))

# Calculate simple interest
SI = (P * R * T) / 100

# Display the result
print("Simple Interest:", SI)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python

# Take input from the user
heads = int(input("Enter total heads: "))
legs = int(input("Enter total legs: "))

# Calculate number of dogs and chickens
dogs = (legs - 2 * heads) // 2
chicken = heads - dogs

# Print the result
print("dogs ->", dogs)
print("chicken ->", chicken)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python

# Take input from the user
n = int(input("Enter the value of n: "))

# Calculate sum of squares
sum_squares = n * (n + 1) * (2 * n + 1) // 6

# Print the result
print("Sum of squares:", sum_squares)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python

# Take input from the user
a = int(input("Enter the first term: "))
b = int(input("Enter the second term: "))
n = int(input("Enter the value of n: "))

# Calculate common difference
d = b - a

# Calculate the nth term
nth_term = a + (n - 1) * d

# Print the result
print("Nth term:", nth_term)
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python

# Input first fraction
num1 = int(input("Enter numerator of first fraction: "))
den1 = int(input("Enter denominator of first fraction: "))

# Input second fraction
num2 = int(input("Enter numerator of second fraction: "))
den2 = int(input("Enter denominator of second fraction: "))

# Calculate sum
numerator = num1 * den2 + num2 * den1
denominator = den1 * den2

# Print the result
print("Sum of fractions:", numerator, "/", denominator)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python

import math

# Input tank dimensions
H = float(input("Enter tank height: "))
L = float(input("Enter tank length: "))
B = float(input("Enter tank breadth: "))

# Input glass dimensions
h = float(input("Enter glass height: "))
r = float(input("Enter glass radius: "))

# Calculate volumes
tank_volume = H * L * B
glass_volume = math.pi * r * r * h

# Calculate number of complete glasses
glasses = int(tank_volume / glass_volume)

print("Number of glasses:", glasses)
````
